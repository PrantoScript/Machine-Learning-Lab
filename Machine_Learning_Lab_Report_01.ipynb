```python
# ============================================================
# MACHINE LEARNING LAB
# Title: Data Preprocessing Using Python
# Dataset: Synthetic Employee Dataset
# ============================================================


# STEP 1: Import Required Libraries

import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler

print("Libraries imported successfully!")


# STEP 2: Upload and Load Dataset

from google.colab import files

uploaded = files.upload()

file_name = next(iter(uploaded))

df = pd.read_csv(file_name)

print("\nDataset Loaded Successfully!")

print("\nFirst Five Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()


# STEP 3: Initial Data Analysis

print("\nMissing Values Before Preprocessing:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nNumerical Data Description:")
print(df.describe())


# STEP 4: Data Cleaning

# Remove duplicate rows
df = df.drop_duplicates()

# Standardize categorical text values
categorical_text_columns = [
    'Department',
    'Education_Level',
    'Location'
]

for column in categorical_text_columns:
    df[column] = (
        df[column]
        .astype('string')
        .str.strip()
        .str.title()
    )

print("\nData Cleaning Completed!")


# STEP 5: Handling Missing Values

# Convert Hire_Date into datetime format
df['Hire_Date'] = pd.to_datetime(
    df['Hire_Date'],
    errors='coerce'
)

# Separate numerical and categorical columns
numerical_columns = df.select_dtypes(
    include=['number']
).columns

categorical_columns = df.select_dtypes(
    include=['object', 'string']
).columns

# Fill numerical missing values using Median
for column in numerical_columns:
    df[column] = df[column].fillna(
        df[column].median()
    )

# Fill categorical missing values using Mode
for column in categorical_columns:
    if df[column].notna().any():
        df[column] = df[column].fillna(
            df[column].mode()[0]
        )

# Fill missing dates using median date
if df['Hire_Date'].notna().any():
    median_date = df['Hire_Date'].dropna().median()
    df['Hire_Date'] = df['Hire_Date'].fillna(median_date)

print("\nMissing Values After Preprocessing:")
print(df.isnull().sum())

print(
    "\nTotal Missing Values:",
    df.isnull().sum().sum()
)


# STEP 6: Feature Extraction from Hire_Date

df['Hire_Year'] = df['Hire_Date'].dt.year
df['Hire_Month'] = df['Hire_Date'].dt.month

# Remove original date column after extraction
df = df.drop(columns=['Hire_Date'])

print("\nDate Feature Extraction Completed!")

print(df[['Hire_Year', 'Hire_Month']].head())


# STEP 7: Encoding Categorical Variables

df = pd.get_dummies(
    df,
    columns=[
        'Department',
        'Education_Level',
        'Location'
    ],
    dtype=int
)

print("\nCategorical Encoding Completed!")

print("\nDataset After Encoding:")
print(df.head())

print("\nTotal Columns After Encoding:", df.shape[1])


# STEP 8: Scaling Numerical Features

numerical_features = [
    'Age',
    'Salary',
    'Years_at_Company',
    'Number_of_Projects',
    'Performance_Score',
    'Hire_Year',
    'Hire_Month'
]

scaler = StandardScaler()

df[numerical_features] = scaler.fit_transform(
    df[numerical_features]
)

print("\nFeature Scaling Completed!")

print("\nDataset After Scaling:")
print(df.head())


# STEP 9: Final Dataset Verification

print("\nFinal Dataset Shape:")
print(df.shape)

print("\nRemaining Missing Values:")
print(df.isnull().sum().sum())

print("\nRemaining Duplicate Rows:")
print(df.duplicated().sum())


# STEP 10: Save Preprocessed Dataset

df.to_csv(
    'preprocessed_employee_dataset.csv',
    index=False
)

print("\nPreprocessed Dataset Saved Successfully!")


# STEP 11: Download Processed Dataset

files.download('preprocessed_employee_dataset.csv')

print("\nAll Preprocessing Tasks Completed Successfully!")
```
